Dataset link: The emails.csv dataset on the Kaggle https://www.kaggle.com/datasets/balaka18/email-spam-classification-dataset-csv



```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn import metrics
```

```python
df=pd.read_csv('emails.csv')
```

# New Section

```python
df.head()
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace = True)
```

```python
df.drop(['Email No.'],axis=1,inplace=True)
X = df.drop(['Prediction'],axis = 1)
y = df['Prediction']
```

```python
from sklearn.preprocessing import scale
X = scale(X)
# split into train and test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 42)
```

##KNN classifier

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=7)

knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)
```

```python
print("Prediction",y_pred)
```

```python
print("KNN accuracy = ",metrics.accuracy_score(y_test,y_pred))
```

```python
print("Confusion matrix",metrics.confusion_matrix(y_test,y_pred))
```

## SVM classifier

```python
# cost C = 1
model = SVC(C = 1)

# fit
model.fit(X_train, y_train)

# predict
y_pred = model.predict(X_test)
```

```python
metrics.confusion_matrix(y_true=y_test, y_pred=y_pred)
```

```python
print("SVM accuracy = ",metrics.accuracy_score(y_test,y_pred))
```

```python
To understand why KNN and SVM are chosen for email spam detection, let's break them down using an intuitive, real-world example.

Imagine every email can be represented by just two features (though real emails have thousands of word counts):
1. Feature X (X₁): Number of times the word "Free" appears in the email.
2. Feature Y (X₂): Number of times the word "Win" appears in the email.
We want our machine learning model to look at these counts and decide whether the email is Normal (Not Spam, 0) or Abnormal (Spam, 1).
1. K-Nearest Neighbors (KNN)
- The Analogy: "Show me your friends, and I’ll tell you who you are."
How it works with our example:
- Suppose you receive a brand-new email that has the word "Free" appearing 4 times and "Win" appearing 3 times (4, 3).
- KNN looks at all the previous emails stored in its database and calculates the mathematical distance (like drawing a straight line on a graph using Euclidean distance) between your new email and all historical emails.
- It finds the K closest emails (let's say K = 5).
- Out of those 5 closest neighbor emails:
  - 4 are labeled as Spam (because they also had high counts of "Free" and "Win").
  - 1 is labeled as Not Spam.
- The Decision: By majority vote, KNN classifies your new email as Spam.
Pros & Cons in Spam Detection:
- Why it's used: It requires zero training time; it just memorizes the dataset and checks distances when a new email arrives.
- Why it struggles here: If your inbox has 50,000 emails, KNN has to calculate the distance to all 50,000 emails every single time a new message arrives, which can become very slow.
2. Support Vector Machine (SVM)
- The Analogy: Drawing a strict fence or wall on a graph that separates two different neighborhoods with the widest possible safe zone on either side.
How it works with our example:
- Plot all your past emails on a 2D graph where the X-axis is "Free" and the Y-axis is "Win".
- You will notice that normal emails cluster tightly near the bottom-left (low counts of "Free" and "Win"), while spam emails scatter toward the top-right (high counts).
- SVM draws a straight line (or a curve / hyperplane) right down the middle to separate the two groups.
- The Magic of SVM (Support Vectors): SVM doesn't care about every single email. It only focuses on the emails that lie right on the edge—the ones closest to the boundary (called Support Vectors). It adjusts the line so that the margin (the empty gap between the boundary line and the closest emails) is as wide as possible.
- The Decision: When a new email arrives at coordinates (4, 3), SVM simply checks which side of the boundary line it falls on. If it falls on the spam side, it blocks it.
Pros & Cons in Spam Detection:
- Why it's used: Text data has thousands of features (every unique word in the dictionary is a dimension). SVM excels brilliantly in high-dimensional spaces and is very memory-efficient during prediction because it only remembers the boundary line and a few support vectors.
- Why it's challenging: Choosing the right kernel or tuning the penalty parameter (C) can require experimentation.
Summary Comparison Table
Feature	KNN (K-Nearest Neighbors)	SVM (Support Vector Machine)
Core Idea	Looks at the closest individual data points.	Finds an optimal boundary line/hyperplane.
Training Phase	Lazy learner (virtually no training time).	Active learner (takes time to calculate the boundary).
Prediction Speed	Slower (computes distances to all points).	Faster (just checks which side of the line the point is on).
Best Suited For	Smaller datasets with clear local clusters.	High-dimensional data like text, emails, and word frequency vectors.
```
